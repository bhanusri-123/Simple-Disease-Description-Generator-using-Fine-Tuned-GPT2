```python
%%writefile /content/nanogpt/data/medical/prepare.py
import os
import pandas as pd
import numpy as np
import tiktoken

data_dir = os.path.dirname(__file__)
csv_path = os.path.join(data_dir, 'disease_dataset.csv')

df = pd.read_csv(csv_path)
df = df.fillna("")
print(f"Original dataset size: {len(df)} rows")

def has_copied_sentence(complex_text, simple_text):
    c_text_lower = str(complex_text).lower()
    s_sentences = [s.strip().lower() for s in str(simple_text).split('.') if len(s.strip()) > 30]
    for s_sent in s_sentences:
        if s_sent in c_text_lower:
            return True
    return False

df['has_copy'] = df.apply(lambda row: has_copied_sentence(row['complex_description'], row['simple_description']), axis=1)
df = df[df['has_copy'] == False]
print(f"Dataset size after nuking copy-pasters: {len(df)} rows")

text = ""
for _, row in df.iterrows():
    ans = str(row['simple_description'])
    if ans.startswith("I'm not sure"):
        ans = "I'm not sure but I think you are asking about something other than a medical condition."
    text += f"Disease: {row['disease']}\n"
    text += f"Complex: {row['complex_description']}\n"
    text += f"Simple: {ans}\n"
    text += "<|endoftext|>\n"

n = len(text)
train_data = text[:int(n*0.9)]
val_data = text[int(n*0.9):]

enc = tiktoken.get_encoding("gpt2")
train_ids = np.array(enc.encode_ordinary(train_data), dtype=np.uint16)
val_ids = np.array(enc.encode_ordinary(val_data), dtype=np.uint16)
train_ids.tofile(os.path.join(data_dir, 'train.bin'))
val_ids.tofile(os.path.join(data_dir, 'val.bin'))
print("Data preparation complete!")
```

```python
import sys; sys.path.insert(0, "src")
from nanogpt.config import parse_cli
from nanogpt.build import build_model
from nanogpt.data import make_get_batch
from nanogpt.trainer import Trainer

cfg = parse_cli([
    "--config", "configs/finetune_gpt2_russell.toml",
    "--dataset=medical",
    "--out_dir=out-finetune-gpt2-medical",
    "--max_iters=300",
])
print(cfg)
```

```python
import os
get_batch = make_get_batch(
    os.path.join(cfg.data_dir, cfg.dataset),
    cfg.block_size,
    cfg.batch_size,
    cfg.device
)
model = build_model(cfg, device=None)
optimizer = model.configure_optimizers(cfg.weight_decay, cfg.learning_rate,
                                         (cfg.beta1, cfg.beta2), device_type="cuda")

trainer = Trainer(model, cfg, get_batch)
trainer.run(optimizer)
```

```python
val_log = trainer.val_log
best_step, best_loss = min(val_log, key=lambda x: x[1])
print(f"Best val loss {best_loss:.4f} occurred at iteration {best_step}")
print(f"trainer.best_val_loss = {trainer.best_val_loss}")
```

```python
import matplotlib.pyplot as plt

train_log = trainer.train_log
val_log = trainer.val_log

def unpack(log):
    steps = [x[0] for x in log]
    losses = [x[1] for x in log]
    return steps, losses

train_steps, train_losses = unpack(train_log)
val_steps, val_losses = unpack(val_log)

plt.figure(figsize=(8,5))
plt.plot(train_steps, train_losses, label="Train Loss", alpha=0.6)
plt.plot(val_steps, val_losses, label="Val Loss", marker='o', linewidth=2)
plt.axvline(x=best_step, color='green', linestyle='--', alpha=0.7, label=f"Best val loss (iter {best_step})")
plt.xlabel("Iteration")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")
plt.legend()
plt.grid(True)
plt.savefig("training_validation_loss.png")
plt.show()
```

```python
import pandas as pd
import difflib
import subprocess
import textwrap

def pretty_print(text, width=80):
    return textwrap.fill(text, width=width)

df = pd.read_csv("data/medical/disease_dataset.csv").fillna("")
valid_diseases = df['disease'].astype(str).str.lower().str.strip().tolist()

def explain_disease(user_disease: str):
    match = df[df['disease'].str.lower().str.strip() == user_disease.lower().strip()]

    if match.empty:
        print(f"\n\033[1mDisease:\033[0m {user_disease}\n")
        print("\033[1mSimple Description:\033[0m\nSorry, I'm not sure. This doesn't appear to be in my dataset.")
        return

    row = match.iloc[0]
    complex_input = row['complex_description']
    dataset_simple = row['simple_description']

    start_text = (
        "### Instruction:\n"
        "Explain the following disease in simple words.\n\n"
        f"### Disease:\n{user_disease}\n\n"
        f"### Complex Description:\n{complex_input}\n\n"
        "### Response:\n"
    )

    result = subprocess.run(
        ["uv", "run", "python", "scripts/sample.py",
         "--out_dir=out-finetune-gpt2-medical",
         f"--start={start_text}",
         "--temperature=0.1",
         "--max_new_tokens=50",
         "--num_samples=1"],
        capture_output=True, text=True
    )

    raw_prediction = result.stdout.split("### Response:")[-1].split("<|endoftext|>")[0].strip()

    # --- Anti-repetition filter: drop sentences that are near-duplicates of ones already kept ---
    sentences = [s.strip() for s in raw_prediction.split('.') if s.strip()]
    unique_sentences = []
    for s in sentences:
        is_duplicate = any(
            difflib.SequenceMatcher(None, s.lower(), kept.lower()).ratio() > 0.60
            for kept in unique_sentences
        )
        if not is_duplicate:
            unique_sentences.append(s)

    # Drop a trailing sentence if it looks cut off mid-way
    if not raw_prediction.endswith(('.', '!', '?')) and len(unique_sentences) > 1:
        unique_sentences = unique_sentences[:-1]

    final_answer = '. '.join(unique_sentences) + ('.' if unique_sentences else '')

    print(f"\n\033[1mDisease:\033[0m {user_disease}\n")
    print(f"\033[1mDataset's Simple Description (Ground Truth):\033[0m\n{pretty_print(dataset_simple)}\n")
    print(f"\033[1mModel's Simple Description:\033[0m\n{pretty_print(final_answer)}")

    return dataset_simple, final_answer

ground_truth, prediction = explain_disease("Common cold")
```

```python
from sentence_transformers import SentenceTransformer
from sklearn.metrics.pairwise import cosine_similarity

model = SentenceTransformer('all-MiniLM-L6-v2')

def sentence_similarity(text1, text2):
    emb1 = model.encode(text1)
    emb2 = model.encode(text2)

    return cosine_similarity([emb1], [emb2])[0][0]

score = sentence_similarity(prediction, ground_truth)
print(f"Sentence-BERT Similarity: {score:.3f}")
```
